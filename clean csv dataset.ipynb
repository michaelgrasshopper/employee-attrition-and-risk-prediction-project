```python
import pandas as pd
import numpy as np

employee_df = pd.read_csv(r'C:\Users\Lenovo\OneDrive\Dokumente\New learning\Ironhack\Final project\WA_Fn-UseC_-HR-Employee-Attrition.csv')

# Data cleaning
# Load dataset if it is not already available in the kernel
if 'employee_df' not in globals():
    employee_df = pd.read_csv(r'C:\Users\Lenovo\OneDrive\Dokumente\New learning\Ironhack\Final project\WA_Fn-UseC_-HR-Employee-Attrition.csv')

# 1) Remove constant or ID-like columns that do not add predictive signal
cols_to_drop = ['EmployeeCount', 'EmployeeNumber', 'Over18', 'StandardHours']
clean_df = employee_df.drop(columns=[c for c in cols_to_drop if c in employee_df.columns], errors='ignore')

# 2) Encode target if needed for classification
clean_df['Attrition'] = clean_df['Attrition'].map({'Yes': 1, 'No': 0}).fillna(clean_df['Attrition']).astype(int)

# 3) Encode categorical features
categorical_cols = clean_df.select_dtypes(include=['object']).columns.tolist()
for col in categorical_cols:
    if col == 'Attrition':
        continue
    clean_df[col] = clean_df[col].astype('category').cat.codes

# 4) Fill missing values
for col in clean_df.columns:
    if clean_df[col].isnull().any():
        if pd.api.types.is_numeric_dtype(clean_df[col]):
            clean_df[col] = clean_df[col].fillna(clean_df[col].median())
        else:
            clean_df[col] = clean_df[col].fillna(clean_df[col].mode()[0])

# 5) Show cleaned summary
clean_df.head()
clean_df.info()
```

```python
from pathlib import Path

source_path = Path(r"C:\Users\Lenovo\OneDrive\Dokumente\New learning\Ironhack\Final project\WA_Fn-UseC_-HR-Employee-Attrition.csv")
output_path = source_path.with_name("clean_employee_data.csv")

clean_df.to_csv(output_path, index=False)

print(f"Saved cleaned data to: {output_path}")
```
